```python
%sql
-- Conversion funnel
SELECT category_code,
  SUM(unique_views) as views,
  SUM(unique_purchases) as purchases,
  ROUND(SUM(unique_purchases)*100.0/SUM(unique_views), 2) as conversion_rate
FROM dev.ecommerce_governed.gold_events
GROUP BY category_code;


```

```python
%sql

-- Customer tiers
SELECT
  CASE WHEN cnt >= 10 THEN 'VIP'
       WHEN cnt >= 5 THEN 'Loyal'
       ELSE 'Regular' END as tier,
  COUNT(*) as customers,
  AVG(total_spent) as avg_ltv
FROM (SELECT user_id, COUNT(*) cnt, SUM(price) total_spent
      FROM dev.ecommerce_governed.silver_events WHERE event_type='purchase' GROUP BY user_id)
GROUP BY tier;
```

```python
%sql
-- Conversion funnel

SELECT category_code,
  SUM(unique_views) as views,
  SUM(unique_purchases) as purchases,
  ROUND(SUM(unique_purchases)*100.0/SUM(unique_views), 2) as conversion_rate
FROM dev.ecommerce_governed.gold_events
GROUP BY category_code
ORDER BY conversion_rate DESC
LIMIT 10;
```
