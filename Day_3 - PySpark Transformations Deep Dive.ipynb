```python
events_oct = spark.read.csv("/Volumes/workspace/ecommerce/ecommerce_data/2019-Oct.csv", header = True, inferSchema=True)
```

```python
events_nov = spark.read.csv("/Volumes/workspace/ecommerce/ecommerce_data/2019-Nov.csv", header = True, inferSchema=True)
```

```python
events_oct.printSchema()
events_nov.printSchema()
```

```python
df_joined_i = events_oct.join(events_nov, events_oct.product_id == events_nov.product_id, 'inner')
df_joined_i.show()
```

```python
events_joined_l = events_oct.join(events_nov, events_oct.product_id == events_nov.product_id, 'left')
events_joined_l.show()
```

```python
events_joined_o = events_oct.join(events_nov, events_oct.product_id == events_nov.product_id, 'outer')
events_joined_o.show()
```

```python
#Calculate running totals
from pyspark.sql.window import Window
from pyspark.sql.functions import sum, col

window = (
    Window
    .partitionBy("user_id")
    .orderBy("event_time")
    .rowsBetween(Window.unboundedPreceding, Window.currentRow)
)
df_total_running_sum = events_oct.withColumn(
    "running_sum_price",
    sum(col("price")).over(window)
)

display(df_total_running_sum)




```
