```python
events_oct = spark.read.csv("/Volumes/workspace/ecommerce/ecommerce_data/2019-Nov.csv", header = True, inferSchema=True)
```

```python
events_oct.show(5)
```

```python
# Basic operations
events_trimmed = events_oct.select("event_type", "category_code", "price")
events_trimmed = events_trimmed.withColumnRenamed("category_code","product_name")

events_trimmed.show(5, truncate=False)
# Filter expensive products
events_trimmed.filter(events_trimmed.price > 1000).show(truncate=False)
#GRoup By
events_trimmed.groupBy("product_name").count().show(truncate=False)
```

```python
#Getting Top 5 Brands
from pyspark.sql.functions import col

top_brands = events_oct.filter(col("brand").isNotNull()).groupBy("brand").count().orderBy("count", ascending=False).limit(5)
display(top_brands)
```
