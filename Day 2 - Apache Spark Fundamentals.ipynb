```python
display(df_movies.describe())
```

```python
df_trimmed = df_movies.select('unit', 'currency')
display(df_trimmed)
df_trimmed.count()
```

```python
df_filtered = df_movies.filter((df_movies.release_year >= 2000) & (df_movies.release_year <= 2010))
display(df_filtered)
```

```python
from pyspark.sql.functions import col

df_filtered = df_movies.filter((col('release_year') >= 2000) & (col('release_year') <= 2010))
display(df_filtered)
df_filtered.count()
```

```python
df_marvel = df_movies.filter(df_movies.studio == 'Marvel Studios')
display(df_marvel)
df_marvel.count()
```

```python
df_industry = df_movies.select("industry").distinct()
display(df_industry)
```

```python
unique_language = df_movies.select("language").distinct()
display(unique_language)
```

```python
df_movies = df_movies.withColumn("profit", col('revenue')-col("budget"))
display(df_movies)
```

```python
df_movies = df_movies.withColumnRenamed("revenue","total_revenue")
display(df_movies)
```
